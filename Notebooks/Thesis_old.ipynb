```python
def get_sample_records(

            file_path,

            number_of_records=5

            ):


    records = []


    with gzip.open(

            file_path,

            'rt',

            encoding='utf-8'

            ) as file:


        for i in range(

                number_of_records

                ):


            record = json.loads(

                        file.readline()

                        )


            records.append(
                    record
                    )


    return records
```

```python
def find_non_empty_values(file_path,
                          column_name):

    with gzip.open(
            file_path,
            'rt',
            encoding='utf-8'
            ) as file:

        for line in file:

            record = json.loads(line)

            value = record.get(column_name)

            if value not in [None, "", [], {}]:

                return value

    return None
```

```python
def count_total_records(file_path):

    total_records = 0

    with gzip.open(
            file_path,
            'rt',
            encoding='utf-8'
            ) as file:

        for line in file:

            total_records += 1

    return total_records
```

```python
def find_unique_values(file_path,
                       column_name,
                       number_of_records=1000):

    unique_values = set()

    with gzip.open(
            file_path,
            'rt',
            encoding='utf-8'
            ) as file:

        for i in range(number_of_records):

            record = json.loads(
                        file.readline()
                        )

            unique_values.add(
                    record[column_name]
                    )

    return unique_values
```

```python
def count_missing_values(file_path,
                         column_name,
                         number_of_records=1000):

    missing_count = 0

    with gzip.open(
            file_path,
            'rt',
            encoding='utf-8'
            ) as file:

        for i in range(number_of_records):

            record = json.loads(
                        file.readline()
                        )

            value = record.get(column_name)

            if value in [None, "", [], {}]:

                missing_count += 1

    return missing_count
```

```python
def search_keyword(file_path,
                   column_name,
                   keyword,
                   number_of_records=1000):

    keyword = keyword.lower()

    with gzip.open(
            file_path,
            "rt",
            encoding="utf-8"
            ) as file:

        for i in range(number_of_records):

            record = json.loads(
                        file.readline()
                        )

            value = record.get(
                        column_name
                        )

            if isinstance(value, str):

                if keyword in value.lower():

                    return value

    return None
```

```python
def get_value_distribution(file_path,
                           column_name,
                           number_of_records=1000):

    distribution = {}

    with gzip.open(
            file_path,
            "rt",
            encoding="utf-8"
            ) as file:

        for i in range(number_of_records):

            record = json.loads(
                        file.readline()
                        )

            value = record.get(
                        column_name
                        )

            distribution[value] = (
                distribution.get(value, 0) + 1
            )

    return distribution
```

```python
def find_most_common_values(file_path,
                            column_name,
                            number_of_records=1000):

    distribution = get_value_distribution(
                        file_path,
                        column_name,
                        number_of_records
                        )

    sorted_values = sorted(
                        distribution.items(),
                        key=lambda item: item[1],
                        reverse=True
                        )

    return sorted_values
```

```python
most_common = find_most_common_values(
                    AMAZON_FASHION_FILE,
                    "rating"
                    )

print(most_common)
```

```python
def filter_records(file_path,
                   column_name,
                   target_value,
                   number_of_records=1000):

    matching_records = []

    with gzip.open(
            file_path,
            "rt",
            encoding="utf-8"
            ) as file:

        for i in range(number_of_records):

            line = file.readline()

            if not line:
                break

            record = json.loads(line)

            value = record.get(column_name)

            if value == target_value:

                matching_records.append(record)

    return matching_records
```

USER PROFILING ENGINE

FUNCTION - 3

--------------------------------------------------

extract_user_intent()

PURPOSE

This function identifies the user's
shopping intent from the current query.

The identified intent helps the AI Personal
Shopping Assistant determine whether
additional contextual information is
required for personalization.

FINAL OUTPUT

USER INTENT

```python
def extract_user_intent(user_profile):

    user_query = user_profile["current_query"].lower()


    TRAVELLING_WORDS = [

        "trip",
        "travel",
        "travelling",
        "vacation",
        "holiday",
        "tour"

    ]


    INTERVIEW_WORDS = [

        "interview",
        "job interview"

    ]


    DATE_WORDS = [

        "date",
        "date night",
        "romantic"

    ]


    GIFT_WORDS = [

        "gift",
        "present",
        "birthday gift"

    ]


    if any(

            word in user_query

            for word in TRAVELLING_WORDS

            ):

        user_profile["intent"] = "Travelling"


    elif any(

            word in user_query

            for word in INTERVIEW_WORDS

            ):

        user_profile["intent"] = "Interview"


    elif any(

            word in user_query

            for word in DATE_WORDS

            ):

        user_profile["intent"] = "Date Night"


    elif any(

            word in user_query

            for word in GIFT_WORDS

            ):

        user_profile["intent"] = "Gift"


    else:

        user_profile["intent"] = "Shopping"


    return user_profile
```

PRODUCT KNOWLEDGE ENGINE

FUNCTION - 3

--------------------------------------------------

calculate_trust_score()

PURPOSE

This function evaluates the trustworthiness
of a product using the available product
information.

The Trust Score assists the Recommendation
Engine in identifying highly reliable
products for personalized recommendations.

FINAL OUTPUT

TRUST SCORE

```python
def calculate_trust_score(product_profile):

    rating = product_profile["rating"]

    review_count = product_profile["review_count"]


    if rating is None or review_count is None:

        return "Not Available"


    if rating >= 4.5 and review_count >= 1000:

        return "Excellent"


    elif rating >= 4.0 and review_count >= 500:

        return "Very Good"


    elif rating >= 3.5:

        return "Good"


    else:

        return "Fair"
```
